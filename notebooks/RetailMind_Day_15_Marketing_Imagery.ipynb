# RetailMind AI — Day 15: Diffusion Marketing Imagery

Generate reproducible product campaign images from catalog metadata with the module in `src/day15_marketing_imagery.py`. The notebook defaults to previewing a prompt and **does not download model weights or generate an image unless you opt in**.

Install the project requirements first. The default Stable Diffusion checkpoint downloads from Hugging Face on first generation; CUDA is preferred and CPU generation may be slow. Check the checkpoint's license before commercial use.

```python
from pathlib import Path
from src.day15_marketing_imagery import (
    DEFAULT_NEGATIVE_PROMPT, DEFAULT_MODEL, DEFAULT_PROMPT_LIBRARY,
    build_marketing_prompt, generate_marketing_image, benchmark_marketing_settings,
)

PRODUCT = {
    'product_name': 'Everyday Tote',
    'category': 'cotton canvas tote bag',
    'style': 'minimal, cheerful, warm colors',
    'setting': 'sunlit city market',
    'brand_voice': 'useful, optimistic, approachable',
    'template': 'lifestyle',
}
SEED = 42
GUIDANCE_SCALE = 7.5
print(f'Model: {DEFAULT_MODEL}')
print(f'Prompt templates: {list(DEFAULT_PROMPT_LIBRARY)}')
```

## 1. Build and review the prompt

Product name, category, visual style, setting, and brand direction form the positive prompt. Inspect it and adjust metadata before generation.

```python
prompt = build_marketing_prompt(**PRODUCT)
print('Positive prompt:\n', prompt)
print('\nNegative prompt:\n', DEFAULT_NEGATIVE_PROMPT)
```

## 2. Generate an image (opt in)

Set `GENERATE_IMAGE = True` to load the checkpoint and run inference. Reusing the same model, seed, and generation settings makes an image reproducible within the normal limits of the runtime and hardware.

```python
GENERATE_IMAGE = False
if GENERATE_IMAGE:
    image, generation = generate_marketing_image(
        **PRODUCT,
        seed=SEED,
        guidance_scale=GUIDANCE_SCALE,
        negative_prompt=DEFAULT_NEGATIVE_PROMPT,
    )
    display(image)
    print(generation['output_path'])
else:
    print('Image generation skipped. Set GENERATE_IMAGE = True to load the model and generate.')
```

## 3. Benchmark inference time (optional)

Text-to-image generation has no ground-truth labels, so accuracy is not applicable. This sweep compares actual inference time and output dimensions across guidance scales on your selected device. It loads the model once and writes `day15_generation_benchmark.csv`.

```python
RUN_BENCHMARK = False
if RUN_BENCHMARK:
    benchmark_rows, benchmark_path = benchmark_marketing_settings(
        **PRODUCT, seed=SEED, guidance_scales=[5.0, 7.5, 10.0]
    )
    print(f'Saved runtime benchmark: {benchmark_path}')
    benchmark_rows
else:
    print('Benchmark skipped. Set RUN_BENCHMARK = True to generate images at each guidance scale.')
```
