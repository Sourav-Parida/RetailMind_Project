Inspect the saved DistilBERT sentiment and review-derived ticket-routing metrics, plus the Day 13 TF-IDF reference. ## Review the saved run

For a fresh fine-tuning run, install the project dependencies (including PyTorch and Transformers), make the review CSV available, run the Day 13 baseline first, then execute `python src/day12_bert_text_intelligence.py`.

```python
from pathlib import Path
import importlib.util
import pandas as pd
from IPython.display import display

ROOT = Path.cwd()
REPORTS = ROOT / 'reports'
```

```python
sentiment_path = REPORTS / 'day12_bert_metrics.csv'
router_path = REPORTS / 'day12_ticket_routing_metrics.csv'
comparison_path = REPORTS / 'day12_bert_comparison.csv'
for path in (sentiment_path, router_path, comparison_path):
    assert path.exists(), f'Missing saved evaluation report: {path}'
sentiment = pd.read_csv(sentiment_path)
router = pd.read_csv(router_path)
comparison = pd.read_csv(comparison_path)
print('Saved sentiment evaluation (held-out split):')
display(sentiment)
print('Saved ticket-routing evaluation (keyword-derived weak labels):')
display(router)
print('Comparison with the Day 13 baseline (separate holdouts):')
display(comparison)
```

## Saved output refresh

This cell displays the saved metrics and figures from `reports/` for this notebook. Training and generation cells are kept intact. Model-heavy workflows could not be rerun here because the required TensorFlow, PyTorch, or Transformers packages are unavailable; opt-in generation cells remain disabled.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, SVG, display

# Re-display the saved metrics and figures for this notebook's project day.
# Set ROOT to the repository root if the notebook is launched elsewhere.
ROOT = Path.cwd()
if ROOT.name == "notebooks": ROOT = ROOT.parent
REPORT_DIR = ROOT / "reports"
DAYS = [12]
print("Showing saved report artifacts for project day(s):", DAYS)
for day in DAYS:
    for path in sorted(REPORT_DIR.glob(f"day{day}*")):
        if path.is_dir():
            for image_path in sorted(path.rglob("*.png")):
                display(Image(filename=str(image_path)))
        elif path.suffix.lower() == ".csv":
            print("\n" + path.name)
            display(pd.read_csv(path, nrows=25))
        elif path.suffix.lower() == ".png":
            display(Image(filename=str(path)))
        elif path.suffix.lower() == ".svg":
            display(SVG(filename=str(path)))
```
